```python
import glob
import pandas as pd

def find_property(csv_folder, **filters):
    """Search the raw CSVs, e.g. find_property(".", **{"Property ID": "DEMO-KPT-8103"})"""
    df = pd.concat([pd.read_csv(f) for f in glob.glob(f"{csv_folder}/*.csv")], ignore_index=True)
    for col, val in filters.items():
        df = df[df[col].astype(str).str.lower() == str(val).lower()]
    return df[["Property ID", "Owner Name", "Area", "Block", "Property Type",
               "House Number", "Street Number", "Price (PKR)"]]

# One specific property by ID
print(find_property(".", **{"Property ID": "DEMO-KPT-8103"}))

# All apartments in FB Area, block 7
print(find_property(".", **{"Area": "FB Area", "Block": 7, "Property Type": "Apartment"}))
```

```python
import os
import sys

import pandas as pd

import sys; sys.path.insert(0, '../backend/ml')  # pipeline module lives in backend/ml
from price_prediction_pipeline import ModelBundle, predict_property_price, resolve_csv_paths
CSV_PATHS = [
    r"apartments.csv", 
    r"houses.csv",  
    r"plots.csv",  
    r"portions.csv", 
    r"commercials.csv", 
]                       
MODEL_PATH = "price_model.joblib" 
def load_raw(csv_paths) -> pd.DataFrame:
    """All raw CSV rows (used for owner lookup and for the menu of valid names)."""
    try:
        files = resolve_csv_paths(csv_paths)
    except FileNotFoundError:
        sys.exit(f"No CSV files found for: {csv_paths}\n"
                 f"Current folder is: {os.getcwd()}\n"
                 "Edit CSV_PATHS at the top of predict.py to point at your CSV folder or files.")
    return pd.concat([pd.read_csv(f) for f in files], ignore_index=True)


def choose(title: str, options: list[str]) -> str:
    """Numbered menu; accepts the number or the name (case-insensitive)."""
    print(f"\n{title}")
    for i, opt in enumerate(options, 1):
        print(f"  {i:>2}. {opt}")
    while True:
        raw = input("Choose number or type name: ").strip()
        if raw.isdigit() and 1 <= int(raw) <= len(options):
            return options[int(raw) - 1]
        hit = [o for o in options if o.lower() == raw.lower()]
        if hit:
            return hit[0]
        print("  Not a valid choice, try again.")


def ask_block() -> int | None:
    """Block is optional: Enter skips it (the model then uses area+type averages)."""
    while True:
        raw = input("\nBlock number (press Enter to skip): ").strip()
        if raw == "":
            return None
        if raw.isdigit():
            return int(raw)
        print("  Please enter a whole number, e.g. 7.")


def show_result(details: dict, result: dict) -> None:
    print("\n" + "=" * 46)
    if details.get("owner"):
        print(f"Owner:            {details['owner']}")
    if details.get("property_id"):
        print(f"Property ID:      {details['property_id']}")
    print(f"Area:             {details['area']}")
    print(f"Block:            {details['block'] if details['block'] is not None else 'not given'}")
    print(f"Property Type:    {details['ptype']}")
    if details.get("listed_price") is not None and not pd.isna(details["listed_price"]):
        print(f"Listed price:     {details['listed_price']:,.0f} PKR")
    print(f"Estimated price:  {result['predicted_price']:,.0f} PKR")
    if result["price_range"]:
        r = result["price_range"]
        print(f"Typical range:    {r['low']:,.0f} to {r['high']:,.0f} PKR")
    print(f"Confidence:       {result['confidence_level']}")
    for w in result["warnings"]:
        print(f"Note:             {w}")
    print("=" * 46)


def main() -> None:
    if not os.path.isfile(MODEL_PATH):
        sys.exit(f"Model file not found: {MODEL_PATH}\n"
                 "Train it first: python price_prediction_pipeline.py <csv_folder> --save price_model.joblib\n"
                 "or edit MODEL_PATH at the top of predict.py.")
    bundle = ModelBundle.load(MODEL_PATH)
    raw = load_raw(CSV_PATHS)

    # Menus show names exactly as written in the CSVs. Only property types that
    # have prices are offered (Commercial has none in the current data).
    priced_types = set(bundle.categories["property_type"])
    areas = sorted(raw["Area"].dropna().unique())
    types = sorted(t for t in raw["Property Type"].dropna().unique() if t.lower() in priced_types)

    while True:
        details = None
        pid = input("\nProperty ID (e.g. DEMO-KPT-8103) or press Enter to enter details manually: ").strip()

        if pid:
            hit = raw[raw["Property ID"].str.lower() == pid.lower()]
            if hit.empty:
                print("  Property ID not found - switching to manual entry.")
            else:
                row = hit.iloc[0]
                details = dict(owner=row["Owner Name"], property_id=row["Property ID"], area=row["Area"],
                               block=int(row["Block"]), ptype=row["Property Type"],
                               listed_price=row["Price (PKR)"])

        if details is None:
            details = dict(area=choose("Area:", areas), ptype=choose("Property type:", types),
                           block=ask_block())

        result = predict_property_price(bundle, area_name=details["area"],
                                        property_type=details["ptype"],
                                        block_number=details["block"])
        show_result(details, result)

        if input("\nAnother property? (y/n): ").strip().lower() != "y":
            break


if __name__ == "__main__":
    main()
```
